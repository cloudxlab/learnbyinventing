# Logarithms

Logarithms sound intimidating, but you already understand them.
You&rsquo;ll prove it to yourself in the next few minutes &mdash;
starting with nothing more than counting zeros.

<div style="font-size:1.5em;">
  <p>In Python, <code>**</code> is the power operator. <code>10**2</code> means
  &ldquo;10 to the power of 2&rdquo;.</p>
  <table style="margin:0.5em auto; font-size:0.95em;">
    <tr><th style="padding:0.3em 0.8em;">Expression</th><th style="padding:0.3em 0.8em;">Value</th><th style="padding:0.3em 0.8em;">Zeros</th></tr>
    <tr><td><code>10**1</code></td><td>10</td><td>1</td></tr>
    <tr><td><code>10**2</code></td><td>100</td><td>2</td></tr>
    <tr><td><code>10**3</code></td><td>1000</td><td>3</td></tr>
    <tr><td><code>10**4</code></td><td>10000</td><td>4</td></tr>
  </table>
  <div class="fragment" data-fragment-index="0">
    <p>The <strong>power</strong> tells you how many zeros.
    The <strong>number of zeros</strong> tells you the power.</p>
    <p>Going from power &rarr; number is <strong>exponentiation</strong>.<br/>
    Going from number &rarr; power is <strong>logarithm</strong>.</p>
  </div>
</div>


<div style="font-size:1.5em;">
  <p>So when someone asks &ldquo;what is log<sub>10</sub>(1000)?&rdquo; they are asking:</p>
  <p style="text-align:center; font-size:1.1em; color:#7a6b3a;">
    <strong>10 to the power of <em>what</em> gives 1000?</strong>
  </p>
  <div class="fragment" data-fragment-index="0">
    <p style="text-align:center; color:#2a7a2a; font-size:1.1em;">Answer: <strong>3</strong>, because 10<sup>3</sup> = 1000.</p>
  </div>
  <div class="fragment" data-fragment-index="1">
    <p>That&rsquo;s it. That&rsquo;s all a logarithm is.</p>

      <tr><td>log<sub>10</sub>(10000)</td><td>10<sup>?</sup> = 10000</td><td>4</td></tr>
    </table>

## Part 2: The Power Question

<div style="font-size:1.5em;">
  <p><strong>Quick fire:</strong> answer without Python.</p>
  <div class="qa-stack">
    <div class="qa-step fragment fade-out" data-fragment-index="0">

    <div class="qa-step fragment fade-in-then-out" data-fragment-index="0">
      <p style="color:#2a7a2a;">&#10003; <strong>2</strong>, because 10<sup>2</sup> = 100.</p>
      <p>10 to the power of <em>what</em> gives <strong>10</strong>?</p>
    </div>
    <div class="qa-step fragment fade-in-then-out" data-fragment-index="1">

      <p>10 to the power of <em>what</em> gives <strong>1</strong>?</p>

  <p><strong>Write <code>my_log(value, base)</code></strong> that tries

<pre><code>def my_log(value, base):
    best_x = 0
    best_diff = abs(value - 1)   # base**0 = 1
    x = 0
    while x &lt;= 20:
        diff = abs(base**x - value)
        if diff &lt; best_diff:
            # update best_x and best_diff
            pass
        x += 0.001
    return round(best_x, 3)</code></pre>
    </div>

## Part 5: The Change of Base Formula

<div style="font-size:1.5em;">
  <p>Python only gives you <code>math.log10()</code> and <code>math.log2()</code> built in.
  What if you need log base 3?</p>
  <div class="fragment" data-fragment-index="0">
    <p>Try this experiment &mdash; compute these ratios:</p>
    <table style="margin:0.5em auto; font-size:0.85em;">
      <tr><th style="padding:0.2em 0.6em;"></th><th style="padding:0.2em 0.6em;">log<sub>10</sub>(value)</th><th style="padding:0.2em 0.6em;">log<sub>10</sub>(base)</th><th style="padding:0.2em 0.6em;">ratio</th><th style="padding:0.2em 0.6em;">expected</th></tr>
      <tr><td>log<sub>3</sub>(27)</td><td>1.431</td><td>0.477</td><td><span class="fragment fade-out" data-fragment-index="1">?</span><span class="fragment" data-fragment-index="1">3.0</span></td><td>3</td></tr>
      <tr><td>log<sub>2</sub>(8)</td><td>0.903</td><td>0.301</td><td><span class="fragment fade-out" data-fragment-index="2">?</span><span class="fragment" data-fragment-index="2">3.0</span></td><td>3</td></tr>
      <tr><td>log<sub>5</sub>(125)</td><td>2.097</td><td>0.699</td><td><span class="fragment fade-out" data-fragment-index="3">?</span><span class="fragment" data-fragment-index="3">3.0</span></td><td>3</td></tr>
    </table>
  </div>
  <div class="qa-stack">
    <p class="qa-step fragment fade-in-then-out" data-fragment-index="1"><strong>Divide log<sub>10</sub>(value) by log<sub>10</sub>(base).</strong> What do you get?</p>
    <p class="qa-step fragment fade-in-then-out" data-fragment-index="2"><strong>Try the next one.</strong> Same pattern?</p>
    <div class="qa-step fragment" data-fragment-index="3">
      <p style="color:#2a7a2a;">&#10003; The ratio always gives the right answer!</p>
    </div>
  </div>
</div>


<div style="font-size:1.5em;">
  <p>You just discovered the <strong>change of base formula</strong>:</p>
  <p style="text-align:center; font-size:1.3em; color:#7a6b3a; margin:0.5em 0;">
    <strong>log<sub>b</sub>(x) = log<sub>10</sub>(x) &divide; log<sub>10</sub>(b)</strong>
  </p>
  <div class="fragment" data-fragment-index="0">
    <p>In Python:</p>
<pre><code>import math
def log_base(value, base):
    return math.log10(value) / math.log10(base)</code></pre>
  </div>
  <div class="fragment" data-fragment-index="1">
    <p style="font-size:0.9em;">
    <code>log_base(27, 3)</code> &rarr; 3.0 &nbsp;&#10003; &nbsp;
    <code>log_base(6, 2)</code> &rarr; 2.585 &nbsp;&#10003; &nbsp;
    <code>log_base(1024, 2)</code> &rarr; 10.0 &nbsp;&#10003;</p>
  </div>
  <div class="fragment" data-fragment-index="2">
    <p>This works with <em>any</em> base in the denominator &mdash;
    log<sub>2</sub>, log<sub>10</sub>, ln &mdash; the ratio is always the same.</p>
    <p style="font-size:0.85em;">(Python&rsquo;s <code>math.log(value, base)</code> does exactly this.)</p>
  </div>
</div>
