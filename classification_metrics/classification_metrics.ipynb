# Classification Metrics

How do you know if a classifier is good? You'll start with raw
predictions, invent the confusion matrix, derive precision, recall,
and F1 — then discover how to choose the best threshold and compare
models using ROC curves and AUC.

    tp, tn, fp, fn = 0, 0, 0, 0
    for a, p in zip(actual, predicted):
        if a == 1 and p == 1: tp += 1
        elif a == 0 and p == 0: tn += 1
        elif a == 0 and p == 1: fp += 1
        elif a == 1 and p == 0: fn += 1
    return tp, tn, fp, fn</code></pre>

## Part 2: Classification Metrics

<div style="font-size:1.3em;">
  <p>Now that we have TP, TN, FP, FN &mdash; what can we measure?</p>
  <table style="margin:0.5em auto; text-align:left;">
    <tr><th style="padding:0.3em 0.5em;">Metric</th><th style="padding:0.3em 0.5em;">Formula</th><th style="padding:0.3em 0.5em;">Intuition</th></tr>
    <tr>
      <td><strong>Precision</strong></td>
      <td>TP / (TP + FP)</td>
      <td>&ldquo;Of everything I called positive, how many really were?&rdquo;</td>
    </tr>
    <tr class="fragment" data-fragment-index="0">
      <td><strong>Recall</strong></td>
      <td>TP / (TP + FN)</td>
      <td>&ldquo;Of all the real positives, how many did I catch?&rdquo;</td>
    </tr>
    <tr class="fragment" data-fragment-index="1">
      <td><strong>F1 Score</strong></td>
      <td>2PR / (P + R)</td>
      <td>Harmonic mean &mdash; balances precision and recall</td>
    </tr>
    <tr class="fragment" data-fragment-index="2">
      <td><strong>Accuracy</strong></td>
      <td>(TP+TN) / Total</td>
      <td>Overall correctness (misleading if classes are imbalanced)</td>
    </tr>
    <tr class="fragment" data-fragment-index="2">
      <td><strong>TPR</strong></td>
      <td>TP / (TP + FN)</td>
      <td>Same as recall &mdash; used in ROC curves</td>
    </tr>
    <tr class="fragment" data-fragment-index="2">
      <td><strong>FPR</strong></td>
      <td>FP / (FP + TN)</td>
      <td>&ldquo;Of all the negatives, how many did I falsely flag?&rdquo;</td>
    </tr>
  </table>
</div>

<div style="font-size:1.3em;">
  <p>The <strong>Area Under the ROC Curve (AUC)</strong> summarizes model quality in a single number:</p>
  <ul style="font-size:0.9em;">
    <li><strong>AUC = 1.0</strong> &rarr; perfect classifier (never makes a mistake)</li>
    <li><strong>AUC = 0.5</strong> &rarr; random guessing (the diagonal line)</li>
    <li><strong>AUC &lt; 0.5</strong> &rarr; worse than random (flip predictions!)</li>
  </ul>

  <div class="fragment" data-fragment-index="1">
<pre><code>def compute_auc(fprs, tprs):
    pairs = sorted(zip(fprs, tprs))
    auc = 0
    for i in range(1, len(pairs)):
        dx = pairs[i][0] - pairs[i-1][0]
        avg_y = (pairs[i][1] + pairs[i-1][1]) / 2
        auc += dx * avg_y
    return auc</code></pre>
    <p style="font-size:0.85em;">The higher the AUC, the better the model separates the two classes &mdash;
    regardless of which threshold you pick.</p>
