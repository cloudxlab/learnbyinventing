  <p>The other way to look at this is &mdash; it is a <strong>linear relationship</strong> between input and output: <strong>y&nbsp;=&nbsp;a&nbsp;&times;&nbsp;x&nbsp;+&nbsp;b</strong>. Let's solve another problem to get a grasp of it.</p>
</div>


<div style="font-size:1.5em;">
  <p>Do you know how an electronic thermometer works?</p>
  <p>When a metal is heated, its resistance changes and that changes the electric current through it. We can measure the current and estimate the temperature from it.</p>
  <div class="fragment" data-fragment-index="0">
    <p>We have the following data points:</p>
    <table style="margin:0.3em auto;">
      <tr><th>Current (I)</th><th>Temperature (T)</th></tr>
      <tr><td>1 mA</td><td>10 &deg;F</td></tr>
      <tr><td>1.2 mA</td><td>50 &deg;F</td></tr>
    </table>
    <p>Find the equation: <strong>T&nbsp;=&nbsp;a&nbsp;&times;&nbsp;I&nbsp;+&nbsp;b</strong>. What are a and b?</p>
  </div>
  <div class="fragment" data-fragment-index="1">
    <p>Put the two values in the equation. You get two equations:</p>
    <p style="font-family:monospace; margin-left:2em;">10 &nbsp;= a &times; 1 &nbsp;&nbsp;+ b &nbsp;&nbsp;&hellip; (1)<br/>50 &nbsp;= a &times; 1.2 + b &nbsp;&nbsp;&hellip; (2)</p>
  </div>
</div>


<div style="font-size:1.5em;">
  <p>Now solve them.</p>
  <div class="fragment" data-fragment-index="0">
    <p style="font-family:monospace; margin-left:2em;">(2) &minus; (1): &nbsp;40 = 0.2a &nbsp;&rarr; &nbsp;<strong>a = 200</strong><br/>From (1): &nbsp;b = 10 &minus; 200 &nbsp;&rarr; &nbsp;<strong>b = &minus;190</strong></p>
  </div>
  <div class="fragment" data-fragment-index="1">
    <p>So, the final equation is:</p>
    <p style="font-size:1.2em; color:#7a6b3a; text-align:center;"><strong>T = 200 &times; I &minus; 190</strong></p>
  </div>
</div>


<div style="font-size:1.5em;">
  <p>You solved two problems by hand. Both times:</p>
  <p style="margin-left:1.5em;">Two data points &rarr; two equations &rarr; solve for <strong>a</strong> and <strong>b</strong>.</p>
  <p>Can you write general Python functions for this?</p>
<pre><code>def fit(x1, y1, x2, y2):
    # return a, b

def predict(a, b, x):
    # return y</code></pre>
</div>


<div style="font-size:1.5em;">
  <p>From y<sub>1</sub> = a &times; x<sub>1</sub> + b &nbsp;and&nbsp; y<sub>2</sub> = a &times; x<sub>2</sub> + b, we get:</p>
<pre><code>def fit(x1, y1, x2, y2):
    a = (y2 - y1) / (x2 - x1)
    b = y1 - a * x1
    return a, b

def predict(a, b, x):
    return a * x + b</code></pre>
  <div class="fragment" data-fragment-index="0">
    <p>Verify with both problems:</p>
    <p style="font-family:monospace; margin-left:2em; font-size:0.85em;">fit(1, 3, 2, 7) &rarr; (4, &minus;1) &nbsp; predict(4, &minus;1, 3) &rarr; 11 &#10003;<br/>fit(1, 10, 1.2, 50) &rarr; (200, &minus;190) &nbsp; predict(200, &minus;190, 1.5) &rarr; 110 &#10003;</p>
  </div>
</div>

<div style="font-size:1.5em;">
  <p>For line y&nbsp;=&nbsp;3x&nbsp;&minus;&nbsp;1, the <strong>residual</strong> at each point is the difference between actual and predicted:</p>
  <table style="margin:0.5em auto;">
    <tr><th>x</th><th>Actual</th><th>Predicted</th><th>Residual</th></tr>
    <tr><td>1</td><td>2</td><td>2</td><td>0</td></tr>
    <tr><td>2</td><td>5</td><td>5</td><td>0</td></tr>
    <tr><td>3</td><td>7</td><td>8</td><td>&minus;1</td></tr>
  </table>
  <p>How do we combine these into a single number?</p>
  <div class="fragment" data-fragment-index="0">
    <p>If we just sum them, positive and negative errors would cancel out! What if we <strong>squared them first</strong>?</p>
  </div>
  <div class="fragment" data-fragment-index="1">
    <p style="text-align:center; font-family:monospace;">(0&sup2; + 0&sup2; + (&minus;1)&sup2;) / 3 = 0.33</p>
    <p>This is the <strong>Mean Squared Error (MSE)</strong>.</p>
  </div>
</div>


<div style="font-size:1.5em;">
  <p>The <strong>best fit</strong> line is the one with the <strong>lowest MSE</strong>. Finding it efficiently is what <em>gradient descent</em> does.</p>
</div>


<div style="font-size:1.5em;">
  <p>Not all patterns are linear. Consider this data:</p>

  <div class="fragment" data-fragment-index="1">

</div>


<div style="font-size:1.5em;">
  <p>There could be more inputs like x<sub>1</sub>, x<sub>2</sub>:</p>

  </div>
</div>


<div style="font-size:1.5em;">
