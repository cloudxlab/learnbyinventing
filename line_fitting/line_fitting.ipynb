  <p>The other way to look at this is &mdash; it is a <strong>linear relationship</strong> between input and output: <strong>y&nbsp;=&nbsp;a&nbsp;&times;&nbsp;x&nbsp;+&nbsp;b</strong>. Let's solve another problem to get a grasp of it.</p>
</div>


<div style="font-size:1.5em;">
  <p>Do you know how an electronic thermometer works?</p>
  <p>When a metal is heated, its resistance changes and that changes the electric current through it. We can measure the current and estimate the temperature from it.</p>
  <div class="fragment" data-fragment-index="0">
    <p>We have the following data points:</p>
    <table style="margin:0.3em auto;">
      <tr><th>Current (I)</th><th>Temperature (T)</th></tr>
      <tr><td>1 mA</td><td>10 &deg;F</td></tr>
      <tr><td>1.2 mA</td><td>50 &deg;F</td></tr>
    </table>
    <p>Find the equation: <strong>T&nbsp;=&nbsp;a&nbsp;&times;&nbsp;I&nbsp;+&nbsp;b</strong>. What are a and b?</p>
  </div>
  <div class="fragment" data-fragment-index="1">
    <p>Put the two values in the equation. You get two equations:</p>
    <p style="font-family:monospace; margin-left:2em;">10 &nbsp;= a &times; 1 &nbsp;&nbsp;+ b &nbsp;&nbsp;&hellip; (1)<br/>50 &nbsp;= a &times; 1.2 + b &nbsp;&nbsp;&hellip; (2)</p>
  </div>
</div>


<div style="font-size:1.5em;">
  <p>Now solve them.</p>
  <div class="fragment" data-fragment-index="0">
    <p style="font-family:monospace; margin-left:2em;">(2) &minus; (1): &nbsp;40 = 0.2a &nbsp;&rarr; &nbsp;<strong>a = 200</strong><br/>From (1): &nbsp;b = 10 &minus; 200 &nbsp;&rarr; &nbsp;<strong>b = &minus;190</strong></p>
  </div>
  <div class="fragment" data-fragment-index="1">
    <p>So, the final equation is:</p>
    <p style="font-size:1.2em; color:#7a6b3a; text-align:center;"><strong>T = 200 &times; I &minus; 190</strong></p>
  </div>
</div>
