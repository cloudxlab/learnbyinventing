if-else, loops, and functions &mdash; all by solving one real problem.

<pre><code>def mysqrt(num, lo, hi, iterations):
    for i in range(iterations):
        guess = (lo + hi) / 2
        if guess * guess > num:
            hi = guess
        else:
            lo = guess

    &mdash; now packaged as a reusable function.</p>
