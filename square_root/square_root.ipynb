### Exercise 2.4 — Let Python Decide

You've been deciding manually: "too high → update hi, too low → update lo."
Python can make this decision for you using `if` and `else`.

Run the code below. Then run the block from `guess = ...` to the last
`print(...)` a few more times. Watch `lo` and `hi` get closer together!


````python
lo = 2
hi = 3
num = 5

guess = (lo + hi) / 2
print(f"Guess: {guess}, squared: {guess * guess}")

if guess * guess > num:
    hi = guess       # guess was too high — search lower half
else:
    lo = guess       # guess was too low — search upper half

print(f"New range: [{lo}, {hi}]")
````

<details><summary>Hint 1</summary>

Copy the block starting from `guess = (lo + hi) / 2` down to `print(f"New range...")` and paste it below itself. Run the whole cell — you'll see two rounds of guessing happen automatically.
</details>

<details><summary>Solution</summary>

```python
lo = 2
hi = 3
num = 5

# Round 1
guess = (lo + hi) / 2
if guess * guess > num:
    hi = guess
else:
    lo = guess
print(f"Round 1: guess = {guess:.6f}, range = [{lo}, {hi}]")

# Round 2
guess = (lo + hi) / 2
if guess * guess > num:
    hi = guess
else:
    lo = guess
print(f"Round 2: guess = {guess:.6f}, range = [{lo}, {hi}]")

# Round 3
guess = (lo + hi) / 2
if guess * guess > num:
    hi = guess
else:
    lo = guess
print(f"Round 3: guess = {guess:.6f}, range = [{lo}, {hi}]")
```

</details>

Each guess cuts your search range in half. After the first guess, you have
half the range. After the second, a quarter. After ten guesses, you've
narrowed it down to $\frac{1}{1024}$ of the original range.

This "halving" strategy is called **binary search** — "binary" because you
split the range in two each time.
